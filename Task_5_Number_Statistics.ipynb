**Level:** 1 · Day 5

### Objective
Practice loops, conditions, the modulo operator, and basic numerical processing by identifying even/odd numbers and finding the largest/smallest values in a list.

### Problem Statement
Create a Python program that identifies even and odd numbers and finds the largest and smallest numbers in a given list.

### Tools Used
- Python
- Jupyter Notebook


```python
numbers = [12, 7, 25, 4, 18, 31, 9, 42, 16, 3]

even_numbers = []
odd_numbers = []

for number in numbers:
    if number % 2 == 0:
        even_numbers.append(number)
    else:
        odd_numbers.append(number)

largest_number = max(numbers)
smallest_number = min(numbers)
```

```python
print("Number Statistics")
print("-" * 30)
print("Original list :", numbers)
print("Even numbers  :", even_numbers)
print("Odd numbers   :", odd_numbers)
print("Largest value :", largest_number)
print("Smallest value:", smallest_number)
```

## Result
- Even numbers: `[12, 4, 18, 42, 16]`
- Odd numbers: `[7, 25, 31, 9, 3]`
- Largest value: `42`
- Smallest value: `3`

The program successfully demonstrates the required concepts.
